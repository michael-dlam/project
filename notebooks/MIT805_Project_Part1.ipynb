```python
from pyspark.sql.functions import col

# Spark session
spark = SparkSession.builder.appName("NYC_Taxi_BigData").getOrCreate()


# Schema normalisation function
def load_and_cast(path):
    df = spark.read.parquet(path)
    return (df
        .withColumn("VendorID", col("VendorID").cast("int"))
        .withColumn("passenger_count", col("passenger_count").cast("int"))
        .withColumn("RatecodeID", col("RatecodeID").cast("int"))
        .withColumn("PULocationID", col("PULocationID").cast("int"))
        .withColumn("DOLocationID", col("DOLocationID").cast("int"))
        .withColumnRenamed("Airport_fee", "airport_fee")
    )
```

```python
# Working dataset (2020–2025) - 5 years
years_working = range(2020, 2026)
months = [f"{m:02d}" for m in range(1, 13)]
working_paths = [f"data/nyc_taxi/yellow_tripdata_{y}-{m}.parquet"
                 for y in years_working for m in months
                 if os.path.exists(f"data/nyc_taxi/yellow_tripdata_{y}-{m}.parquet")]

working_df = spark.read.parquet(*working_paths)
row_count_working = working_df.count()
```

```python
# Expanded size estimate
sample_row = working_df.limit(1).toPandas().iloc[0]
avg_row_size = sample_row.astype(str).apply(len).sum()
expanded_size_working = (row_count_working * avg_row_size) / (1024**3)

print(f"Working dataset rows: {row_count_working:,}")
print(f"Estimated expanded size: {expanded_size_working:.2f} GB")
```

```python
# 4. Processing dataset (2024–2025, normalized + cleaned) - 2 years
years_processing = range(2024, 2026)
paths_processing = [f"data/nyc_taxi/yellow_tripdata_{y}-{m}.parquet"
                    for y in years_processing for m in months
                    if os.path.exists(f"data/nyc_taxi/yellow_tripdata_{y}-{m}.parquet")]

dfs = [load_and_cast(p) for p in paths_processing]
processing_df = dfs[0]
for df in dfs[1:]:
    processing_df = processing_df.unionByName(df, allowMissingColumns=True)
```

```python
# Cleaning

# Cleaning with year filter included
from pyspark.sql.functions import year

# Cleaning with year filter included
processing_df = (
    processing_df
    .filter((col("tpep_pickup_datetime").isNotNull()) & (col("tpep_dropoff_datetime").isNotNull()))
    .filter(col("fare_amount") > 0)
    .dropDuplicates()
    .filter(col("trip_distance") < 100)
    .filter((year(col("tpep_pickup_datetime")) >= 2024) & (year(col("tpep_pickup_datetime")) <= 2025))
)


# Row count after cleaning
row_count_processing = processing_df.count()

# Expanded size estimate after cleaning
sample_row = processing_df.limit(1).toPandas().iloc[0]
avg_row_size = sample_row.astype(str).apply(len).sum()
expanded_size_processing = (row_count_processing * avg_row_size) / (1024**3)

print(f"Processing dataset rows after cleaning: {row_count_processing:,}")
print(f"Estimated expanded size after cleaning: {expanded_size_processing:.2f} GB")
```

```python
from pyspark.sql.functions import col, when, avg, count, month, year

# Exploratory Data Analysis

# Trip categories
processing_df = processing_df.withColumn(
    "trip_category",
    when(col("trip_distance") < 2, "short")
    .when((col("trip_distance") >= 2) & (col("trip_distance") < 10), "medium")
    .otherwise("long")
)

fare_stats = (processing_df
              .groupBy("trip_category")
              .agg(avg("fare_amount").alias("avg_fare"),
                   count("*").alias("trip_count"))
              .orderBy("trip_category"))
fare_stats.show()

# Payment type distribution
payment_stats = (processing_df
                 .groupBy("payment_type")
                 .agg(count("*").alias("count"))
                 .orderBy("count", ascending=False))
payment_stats.show()

# Monthly trends
monthly_trends = (processing_df
                  .groupBy(year("tpep_pickup_datetime").alias("year"),
                           month("tpep_pickup_datetime").alias("month"))
                  .agg(count("*").alias("trip_count"),
                       avg("fare_amount").alias("avg_fare"))
                  .orderBy("year", "month"))
monthly_trends.show()
```

```python
# Summary statistics for key numeric fields
processing_df.describe(["trip_distance", "fare_amount", "tip_amount", "total_amount"]).show()
```

```python
# Correlations (sample for efficiency)
sample_pdf = processing_df.sample(fraction=0.001, seed=42).toPandas()
print("Correlation matrix (sampled):")
print(sample_pdf[["trip_distance", "fare_amount", "tip_amount", "total_amount"]].corr())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram of trip distances
plt.figure(figsize=(8,5))
sns.histplot(sample_pdf["trip_distance"], bins=50, kde=True)
plt.title("Distribution of Trip Distances")
plt.xlabel("Miles")
plt.ylabel("Frequency")
plt.show()

# Scatter plot: distance vs fare
plt.figure(figsize=(8,5))
sns.scatterplot(x="trip_distance", y="fare_amount", data=sample_pdf, alpha=0.3)
plt.title("Trip Distance vs Fare Amount")
plt.xlabel("Trip Distance (miles)")
plt.ylabel("Fare Amount ($)")
plt.show()

# Boxplot to highlight anomalies in fares
plt.figure(figsize=(8,5))
sns.boxplot(x=sample_pdf["fare_amount"])
plt.title("Fare Amount Anomalies (Boxplot)")
plt.xlabel("Fare Amount ($)")
plt.show()
```

```python
# Payment type distribution
payment_stats_pdf = (processing_df
                     .groupBy("payment_type")
                     .agg(count("*").alias("count"))
                     .orderBy("count", ascending=False)
                     .toPandas())

plt.figure(figsize=(6,4))
sns.barplot(x="payment_type", y="count", data=payment_stats_pdf)
plt.title("Payment Type Distribution")
plt.xlabel("Payment Type Code")
plt.ylabel("Count")
plt.show()
```

```python
# Monthly trends
monthly_pdf = (processing_df
               .groupBy(year("tpep_pickup_datetime").alias("year"),
                        month("tpep_pickup_datetime").alias("month"))
               .agg(count("*").alias("trip_count"),
                    avg("fare_amount").alias("avg_fare"))
               .orderBy("year", "month")
               .toPandas())

plt.figure(figsize=(10,6))
sns.lineplot(x="month", y="avg_fare", hue="year", data=monthly_pdf, marker="o")
plt.title("Monthly Average Fare Trends (2024–2025)")
plt.xlabel("Month")
plt.ylabel("Average Fare ($)")
plt.show()
```
